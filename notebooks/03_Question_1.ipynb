```python
import pandas as pd
import numpy as np
```

```python
cols = [
    # Adresse
    "adresse_ban",

    ## Structure logement & immeuble :
    "type_batiment",
    "annee_construction",
    "periode_construction",
    "nombre_appartement",
    "surface_habitable_desservie_par_installation_ecs_n1",
    "surface_chauffee_installation_chauffage_n1",
    "surface_chauffee_installation_chauffage_n2",
    "surface_climatisee",

    # DPE
    "etiquette_dpe",
    "etiquette_ges",
    "date_etablissement_dpe",
    "date_derniere_modification_dpe",

    # Chauffage électrique
    "conso_chauffage_ef_energie_n1",
    "conso_chauffage_ef_energie_n2",
    "conso_chauffage_ef_energie_n3",
    "conso_chauffage_ep",
    "type_energie_n1",
    "type_energie_n2",
    "type_energie_n3",

    # ECS électrique
    "conso_ecs_ef_energie_n1",
    "conso_ecs_ef_energie_n2",
    "conso_ecs_ef_energie_n3",

    # Éclairage
    "conso_eclairage_ef",

    # Auxiliaires
    "conso_auxiliaires_ef",

    # Refroidissement
    "conso_refroidissement_ef",

    # Coût 
    "cout_chauffage_energie_n1",
    "cout_chauffage_energie_n2",
    "cout_chauffage_energie_n3",
    "cout_ecs_energie_n1",
    "cout_ecs_energie_n2",
    "cout_ecs_energie_n3",
    "cout_eclairage",
    "cout_auxiliaires",
    "cout_refroidissement"
]
```

```python
df_dpe_75 = pd.read_csv("../data/dpe/dpe_paris_full.csv", usecols=cols, low_memory=False)
df_dpe_75.head()
```

```python
df_dpe_75["conso_chauffage_elec"] = (
    np.where(df_dpe_75["type_energie_n1"] == "Électricité", df_dpe_75["conso_chauffage_ef_energie_n1"], 0) +
    np.where(df_dpe_75["type_energie_n2"] == "Électricité", df_dpe_75["conso_chauffage_ef_energie_n2"], 0) +
    np.where(df_dpe_75["type_energie_n3"] == "Électricité", df_dpe_75["conso_chauffage_ef_energie_n3"], 0)
)
```

```python
df_dpe_75["conso_ecs_elec"] = (
    np.where(df_dpe_75["type_energie_n1"] == "Électricité", df_dpe_75["conso_ecs_ef_energie_n1"], 0) +
    np.where(df_dpe_75["type_energie_n2"] == "Électricité", df_dpe_75["conso_ecs_ef_energie_n2"], 0) +
    np.where(df_dpe_75["type_energie_n3"] == "Électricité", df_dpe_75["conso_ecs_ef_energie_n3"], 0)
)
```

```python
df_dpe_75["conso_eclairage_elec"] = df_dpe_75["conso_eclairage_ef"].fillna(0)
df_dpe_75["conso_auxiliaires_elec"] = df_dpe_75["conso_auxiliaires_ef"].fillna(0)
df_dpe_75["conso_refroidissement_elec"] = df_dpe_75["conso_refroidissement_ef"].fillna(0)
```

```python
df_dpe_75["totale_conso_electrique"] = (
    df_dpe_75["conso_chauffage_elec"] +
    df_dpe_75["conso_ecs_elec"] +
    df_dpe_75["conso_eclairage_elec"] +
    df_dpe_75["conso_auxiliaires_elec"] +
    df_dpe_75["conso_refroidissement_elec"]
)
```

```python
df_dpe_75["cout_chauffage_elec"] = (
    np.where(df_dpe_75["type_energie_n1"] == "Électricité", df_dpe_75["cout_chauffage_energie_n1"], 0) +
    np.where(df_dpe_75["type_energie_n2"] == "Électricité", df_dpe_75["cout_chauffage_energie_n2"], 0) +
    np.where(df_dpe_75["type_energie_n3"] == "Électricité", df_dpe_75["cout_chauffage_energie_n3"], 0)
)
```

```python
df_dpe_75["cout_ecs_elec"] = (
    np.where(df_dpe_75["type_energie_n1"] == "Électricité", df_dpe_75["cout_ecs_energie_n1"], 0) +
    np.where(df_dpe_75["type_energie_n2"] == "Électricité", df_dpe_75["cout_ecs_energie_n2"], 0) +
    np.where(df_dpe_75["type_energie_n3"] == "Électricité", df_dpe_75["cout_ecs_energie_n3"], 0)
)
```

```python
df_dpe_75["cout_eclairage_elec"] = df_dpe_75["cout_eclairage"].fillna(0)
df_dpe_75["cout_auxiliaires_elec"] = df_dpe_75["cout_auxiliaires"].fillna(0)
df_dpe_75["cout_refroidissement_elec"] = df_dpe_75["cout_refroidissement"].fillna(0)
```

```python
df_dpe_75["totale_cout_electrique"] = (
    df_dpe_75["cout_chauffage_elec"]
    + df_dpe_75["cout_ecs_elec"]
    + df_dpe_75["cout_eclairage_elec"]
    + df_dpe_75["cout_auxiliaires_elec"]
    + df_dpe_75["cout_refroidissement_elec"]
)
```
