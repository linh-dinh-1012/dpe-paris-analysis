```python
import requests
import pandas as pd
from tqdm import tqdm
```

```python
# Get DPE data from ADEME source  
base_url = "https://data.ademe.fr/data-fair/api/v1/datasets/dpe03existant/lines"

def fetch_paris(size=1000):
    url = base_url
    params = {
        "qs": 'nom_commune_ban:"Paris"',
        "size": size
    }

    dfs = []

    # Request to get the sum:
    r = requests.get(url, params=params).json()
    total = r.get("total", None)
    print("Total Paris:", total)

    pbar = tqdm(total=total)

    while True:

        # fetch page
        response = requests.get(url, params=params).json()
        results = response.get("results", [])

        if not results:
            break

        dfs.append(pd.DataFrame(results))
        pbar.update(len(results))

        # pagination with NEXT URL
        next_url = response.get("next")
        if not next_url:
            break

        # move to the next_url
        url = next_url
        params = {}  # next URL contains query + after token

    pbar.close()
    return pd.concat(dfs, ignore_index=True)


df_paris = fetch_paris()
df_paris.to_csv("dpe_paris.csv", index=False)
df_paris.head()
```
